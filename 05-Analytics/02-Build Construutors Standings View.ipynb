### Build Constructors Standings 

```sql
SELECT * FROM formula1.gold.dim_constructors;
SELECT * FROM formula1.gold.fact_session_results
```

**Requirements**
Creating view that supports Constructors standings and total_points, shows race_starts, number_of_wins, number_of_podium. 
constructors recods should be grouped by season, constructor_id, constructor_name, constructor_nationality, 
Ranked () Partition by season and  order by total_points, number_of_wins

```sql
CREATE OR REPLACE VIEW formula1.gold.v_constructor_standings AS
WITH constructor_summary AS 
(
  SELECT
  r.season,
  c.name AS constructor_name,
  c.constructor_id,
  COUNT_IF(r.has_points) AS race_starts,
  SUM(r.points) AS total_points,
  COUNT_IF(r.is_win) AS number_of_wins,
  COUNT_IF(r.has_points) AS number_of_podium
FROM
  formula1.gold.dim_constructors AS c
    JOIN formula1.gold.fact_session_results AS r
      ON c.constructor_id = r.constructor_id
GROUP BY
  r.season,
  c.constructor_id,
  c.name,
  c.nationality
  )

SELECT 
  season,
  constructor_name,
  Rank() OVER (PARTITION BY season ORDER BY total_points DESC, number_of_wins DESC) As Standing, 
  constructor_id,
  race_starts,
  total_points,
  number_of_wins,
  number_of_podium
 FROM constructor_summary
```

```sql
SELECT * FROM formula1.gold.v_constructor_standings
```
