### Build Races Dimension

```python
%run ../00-common/01-environment-config
```

```python
circuits_table = spark.table(f"{catalog_name}.{silver_schema}.circuits")
races_table = spark.table(f"{catalog_name}.{silver_schema}.races")
table_name = f"{catalog_name}.{gold_schema}.dim_races"
```

```python
display(circuits_table)
display(races_table)
```

### Joining Two Tablesa and Keeping Required Columns

```python
races_df = (
    circuits_table
    .join(races_table,
           circuits_table.circuit_id == races_table.circuit_id,
            "inner",)
    .select(
        races_table.season,
        races_table.round,
        races_table.race_name,
        races_table.date.alias("race_date"),
        circuits_table.circuit_name,
        circuits_table.locality,
        circuits_table.country
    )
)
```

```python
display(races_df)
```

### Writing data to gold table

```python
(
    races_df
    .write
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .format("delta")
    .saveAsTable(table_name)
    )
display(spark.table(table_name))
```
