## Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS formula1_external_location
URL 'abfss://formula1@storageaccoun1t.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL formula1_storage_credential)
COMMENT 'External location for Formula 1 container';
```

### Create Catalog Formula1

```sql
SHOW CATALOGS
```

```sql
CREATE CATALOG IF NOT EXISTS formula1
MANAGED LOCATION 'abfss://formula1@storageaccoun1t.dfs.core.windows.net/'
COMMENT 'Main catalog for Formula1 project'
```

### Create Required Schemas 

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
MANAGED LOCATION 'abfss://formula1@storageaccoun1t.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.silver
MANAGED LOCATION 'abfss://formula1@storageaccoun1t.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.gold
MANAGED LOCATION 'abfss://formula1@storageaccoun1t.dfs.core.windows.net/bronze';

```
