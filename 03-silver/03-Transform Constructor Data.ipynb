### Transform Constructor Data

```python
%run ../00-common/01-environment-config
```

```python
bronze_table = f"{catalog_name}.{bronze_schema}.constructors"
silver_table = f"{catalog_name}.{silver_schema}.constructors"
```

#### 1. Read Constructors data

```python
constructors_df = spark.read.table(bronze_table)
display(constructors_df)
```

#### 2. `Check` and remove NUll values

```python
from pyspark.sql import functions as F

null_counts_df = constructors_df.select([
    F.sum(F.col(column).isNull().cast("int")).alias(column)
    for column in constructors_df.columns
])

display(null_counts_df)
```

#### 3. Drop Unncessesory columns

```python
constructors_selected_df = constructors_df.drop("url")
display(constructors_selected_df)
```

#### 4. Rename Columns

```python
constructors_renamed_df = constructors_selected_df.withColumnsRenamed({
    "constructorId": "constructor_id",
})
```

#### 5. Check for duplicates and Remove if needed

```python
constructors_distinct_df = constructors_renamed_df.distinct()
constructors_distinct_df.display()
```

### 6. Transform Values of Columns nationality, name to Title Case

```python
constructors_final_df = constructors_distinct_df.withColumns({
    "name": F.initcap(F.col("name")),
    "nationality": F.initcap(F.col("nationality"))
})

display(constructors_final_df)
```

### 7. Writing data to bronze table

```python
(
    constructors_final_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(silver_table)
)
```
